Hva ønsker vi å analysere?

```python
years = ["20-21", "21-22", "22-23", "23-24", "24-25"]

for year in years:
    G = data[f"{year}.SE.G.SK"].mean()
    J = data[f"{year}.SE.J.SK"].mean()
    print(f"{year}: Boys={G:.2f}, Girls={J:.2f}")
```
